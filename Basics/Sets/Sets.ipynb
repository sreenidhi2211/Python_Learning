# **WHAT IS SET**


*   **unordered**
*   **immutable,unique**
*   as it is unordered there is no indexing






# CREATING SETS

```python
# using normal flower brackets
set_int={1,2,3,4,5}
print(set_int)
set_str={'a','b','c','d'}
print(set_str)
#creating an empty set
# use set(argument) function:only argument allowed and you can use lists,tuples or even sets
set_fun=set([1,3,2,1,5,4])
print(set_fun)
# using empty parenthesis creates a empty dictonary

```

# ACCESSING THE SET ELEMENT
*   we cannot access using index
*   so we check using for loop and in keyword



```python
for i in set_int:
  print(i)
print(7 if 7 in set_int else "not present")
```

# **MODIFYING THE SET**
> **1.** **ADD ELEMENTS TO THE SET**

    *   .add():you add a single element using this method
    *   .update():you can add a sequence data type i.e multiple elements


> **2.**  **REMOVE ELEMENTS FROM THE SET**

    *   .remove():throws key error if element is not present
    *   .discard():throws no such error
    *   .clear():emptys the set amd return empty set   
    * .del(): empties the set deletes the set object and throws Name error when we try to access the set that is deleted





```python
# ADD ELEMENTS TO THE SET
set_2={3,4,7,2,31,78}
set_2.add("abcd")
print(set_2)
set_2.add(67.98)
print(set_2)
set_2.update((2,'k',56.5))
print(set_2)
set_2.update({2,3,'discard',True})
print(set_2)
```

```python
# REMOVE ELEMENTS FROM THE LIST
#.remove()
set_3={'a',123,76.56,False,(12,3,4)}
print(set_3)
set_3.remove(False)
print(set_3)
# element that is not in the set
#set_3.remove(True)
"""
KeyError: True
"""
#.discard()
set_3.discard((12,3,4))
print(set_3)
#element that is not in the set
set_3.discard(False)
print(set_3)
#.clear
set_3.clear()
print(set_3)
#.del
del set_3
# print(set_3)
"""
NameError: name 'set_3' is not defined
"""



```

# **SET OPERATIONS**

> **1. UNION**


* `.union()` : Combines all unique elements from two or more sets.
* `|` : Performs the union operation and returns all unique elements.


> **2. INTERSECTION**


* `.intersection()` : Returns only the elements common to the given sets.
* `&` : Performs the intersection operation and returns common elements.


> **3. DIFFERENCE**

* `.difference()` : Returns elements present in the first set but not in the other set.
* `-` : Performs the difference operation and returns non-common elements.


> **4. SYMMETRIC DIFFERENCE**


* `.symmetric_difference()` : Returns elements present in either set but not in both sets.
* `^` : Performs the symmetric difference operation.

> **5. SUBSET CHECK**

* `.issubset()` : Checks whether all elements of one set are present in another set.
* `<=` : Performs the subset check operation.



> **6. SUPERSET CHECK**


* `.issuperset()` : Checks whether a set contains all elements of another set.
* `>=` : Performs the superset check operation.



> **7. DISJOINT CHECK**


* `.isdisjoint()` : Checks whether two sets have no common elements.

```python
#Union
set_u1={1,2,3,4}
set_u2={5,6,7,8}
set_union=set_u1.union(set_u2)
#or set_union=set_u1|set_2
print(set_union)
set_intersection=set_u1.intersection(set_u2)
#or set_intersection=set_u1 & set_u2
print(set_intersection)
# Difference
set_diff=set_u1.difference(set_u2)
# os set_diff=set_u1-set_u2
print(set_diff)
#Symmetric difference
set_sym_diff=set_u1.symmetric_difference(set_u2)
print(set_sym_diff)
# or set_sym_diff=set_u1^set_u2


```

# SUBSETS & SUPERSET OPERATIONS

```python
# subset :returns true or false
set_s1={1,2,3,4,5}
set_s2={3,4,5}
print(set_s1.issubset(set_s2))
print(set_s2<=set_s1) # using symbols
#proper subset
print(set_s2<set_s1)
# superset
print(set_s1.issuperset(set_s2))
print(set_s2>set_s1)
```
