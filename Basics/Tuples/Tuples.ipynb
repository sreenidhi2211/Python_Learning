# WHAT IS A TUPLE
* a sequence data type that is:
1. Ordered,**Immutable**
2. Dynamic data Types
3. Can be empty

# CREATING A TUPLE

```python
#empty tuple
empty_tuple=()
print(type(empty_tuple))
#integer tuple
int_tuple=(1,2,3,4,5)
print(int_tuple)
print(type(int_tuple))
#mixed tuple
mixed_tuple=(1,'rest',[2,4,6],{2.3,4.5,6.7},True)
print(mixed_tuple)
print(type(mixed_tuple))
#using method to create a tuple by giving any datatype as a argument
lst=['r','t','s']
print(type(lst))
method_tuple=tuple(lst)# as a method it expects only one argument
print(method_tuple)
print(type(method_tuple))
# tuple to list again
# method_tuple=list(method_tuple)
# print(type(method_tuple))
```

# ITERATING THROUGH A TUPLE


```python
for i in int_tuple:
  print(i)
```

# ACCESSING ELEMENTS IN TUPLE
* same as lists

```python
t_1=(2,4,1,6,5)
#single indexing
print(t_1[-1])
print(t_1[0])
#slicing
print(t_1[1:])
print(t_1[1:3])
print(t_1[0::])
#reverse the tuple
print(t_1[::-1])
```

```python
print(mixed_tuple*3)
print(int_tuple *3)
```

# PACKING AND UNPACKING

```python
#packing a tuple
packing_tuple=1,2,3
print(packing_tuple)
# unpacking a tuple
a,b,c=packing_tuple
print(a)
print(b)
print(c)
# unpacking with a *
packed_tup=(1,2,3,4,5,6,7)
f,*m,l=packed_tup
print(f)
print(m)# the middle elements all are packed into a list
print(l)
```

# NESTED TUPLES

```python
nest_tup=((1,2,3),('r','j','m'))
print(nest_tup)
print("the first tuple is: ",nest_tup[0])
print("the second tuples 2,3 element  is: ",nest_tup[1][1:])
# If a mutable item like list is present in the tuple you can make changes in that element only
tuple_change=(1,2,3,[45,87,34,107],"String")
tuple_change[3][3]=999
print(tuple_change)
```

# ITERATING THROUGH NEST TUPLES


```python
for sub_t in nest_tup:
  for e_t in sub_t:
    print(e_t,end=' ')
  print()
```

# COPYING A TUPLE

```python
tuple1 = (0, 1, 2, 3, 4, 5)
tuple2 = tuple1
tuple2=list(tuple2)
tuple2.append(6)
tuple2=tuple(tuple2)
print(tuple2)
print(tuple1)
```
