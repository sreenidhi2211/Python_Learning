## Nested Tuple Flattening
Write a recursive Python function to flatten a deeply nested tuple of tuples into a single flat tuple containing all the individual values.

```python
n_tuple=((1,2), (2, 3), (4, 5, 6, 7))
flattened=[]
for x in n_tuple:
  for y in x:
    flattened.append(y)

print(tuple(flattened))

# suppose a nested tuple is inside a nest tuple
nested_tuple=(1, (2, 3), (4, (5, (6, 7))))
# the above program throughs error
# in such cases we have isinstance(item,tuple)
# what it does: if the current item in the iterable is tuple it dives into the item else return(yields) the value
def flattened(nested_tuple):
  for item in nested_tuple:
    if isinstance(item,tuple):
      yield from flattened(item)
    else:
      yield item

result=tuple(flattened(nested_tuple))
print(result)
```
