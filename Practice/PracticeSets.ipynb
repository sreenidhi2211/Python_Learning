# **Two friends made lists of movies they've watched**



```

raj_movies = {"Inception", "Titanic", "Avatar"}
sam_movies = {"Titanic", "Avatar", "Interstellar"}

```

Find:
* a) Movies both have watched
* b) Movies only Raj has watched
* c) All movies watched by each of them but not both (no duplicates)

````python
raj_movies = {"Inception", "Titanic", "Avatar"}
sam_movies = {"Titanic", "Avatar", "Interstellar"}
# watched by both
print(f"Movies watched by both are : {raj_movies.intersection(sam_movies)}")
# by only raj
print(f"Movies watched by raj are : {raj_movies.difference(sam_movies)}")
# by either
print(f"Movies watched by either are : {raj_movies.symmetric_difference(sam_movies)}")
````

# **Find the string has Unqiue characters or not**
* Write a function that takes a string and returns True if it has no repeated characters, using a set.
* Example: "abc" → True, "hello" → False.

````python
def is_unique(st):
  # st = list(st)
  # st_set = set(st)
  # st_set = list(st_set)
  if len(st) == len(set(st)):
    return True
  else:
    return False

st = input("Write a string or a word : ")
print(is_unique(st))

````

# **String with all vowels**
* Write a function has_all_vowels(s) that returns True if the string s contains all of a, e, i, o, u at least once (case-insensitive).
* Use set operations — no manual loops checking each vowel individually.

````python
vowels = {'a','e','i','o','u'}
def all_vowels(s):
  s =set(s)
  return vowels.issubset(s)

s = input("Enter the string: ").lower()
print("Does the string contain all the vowels ? ")
print("Yes" if all_vowels(s) else "No")
````

#**Unique Intersection Only**
* Given two lists representing two people's shopping carts, find items that are common to both carts but appear only once in each (i.e., ignore items that appear multiple times in either cart).



```
cart1 = ["milk", "bread", "milk", "eggs"]
cart2 = ["bread", "eggs", "eggs", "butter"]
```

````python
cart1 = ["milk", "bread", "milk", "eggs"]
cart2 = ["bread", "eggs", "eggs", "butter"]
cart1_unique = {x for x in cart1 if cart1.count(x) == 1} # includes the items that appear in cart1 only once
cart2_unique = {x for x in cart2 if cart2.count(x) == 1} # includes the items that appear in cart2 only once
print(cart1_unique.intersection(cart2_unique)) # intersection to find the intersection
````
